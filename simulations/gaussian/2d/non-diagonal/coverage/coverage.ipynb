### Compute empirical coverage intervals for $\Theta$ in the Brownian non-diagonal case

* for each Monte Carlo sample, compute a 95 % confidence interval
* Check if true $\Theta$ is within that interval
* Divide number of $\Theta$ contained with number of Monte Carlo samples (100)

Parameters

* n = 25000 number of observations
* T = 1000 np.linspace(100, 4000, 5, dtype=int) # length of time series

* dim = 2 dimension - number of interest rates to simulate
* num_it = 50 number of iterations - how many paths are simulated

* H = np.array([0.5, 0.5]) Hurst indices for components

* b = np.matrix([[1],[3]])  long term mean
* theta = np.matrix([[0.5, 0.1], [0.1, 0.3]])  mean reversion + interactions
* sigma = np.diag([1,2]) # volatility

* s = int(5/T * n)  upper bound for integrals

* 100 monte carlo samples

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pickle
```

```python
filename = "theta_hats_monte_carlo_non_diagonal.pkl"
with open(f"{filename}", "rb") as f:
    theta_hats = pickle.load(f)
```

```python
z = 1.96
n = len(theta_hats[0])
theta = np.matrix([[0.5, 0.1], [0.1, 0.3]])
bottomright = []
bottomconf = []
upperconf = []
covered = np.zeros((len(theta_hats), 2, 2))
for round in range(len(theta_hats)):
    m = np.mean(theta_hats[round], axis=0)
    s = np.std(theta_hats[round], axis=0)
    conf_lower = m - z*s/np.sqrt(n)
    conf_upper = m + z*s/np.sqrt(n)

    bottomconf.append(conf_upper[1, 1])
    upperconf.append(conf_upper[0,0])

    covered[round, :, :] = np.logical_and(theta >= conf_lower, theta <= conf_upper)
    for j in theta_hats[round]:
        bottomright.append(j[1, 1])

```

```python
np.sum(covered, axis=0)/100
```
